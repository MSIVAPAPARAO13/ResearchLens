# 01. PDF Processing & Ingestion Exploration

This notebook demonstrates how ResearchLens ingests raw academic PDFs, extracts per-page text, and parses document metadata.

```python
import fitz

from researchlens.ingestion.pdf_loader import load_pdf

# Create an in-memory test document
doc = fitz.open()
p1 = doc.new_page()
p1.insert_text((50, 72), "Abstract\nRecent developments in transformer architectures...")
pdf_bytes = doc.tobytes()
doc.close()

document = load_pdf(pdf_bytes, filename="demo_paper.pdf")
print(f"Loaded: {document.filename}, Pages: {document.page_count}, Words: {document.word_count}")
```
