# 02. Section Analysis & Noise Preprocessing

This notebook demonstrates regex-based academic section boundary detection, noise cleaning, and sentence filtering.

```python
from researchlens.preprocessing.cleaner import clean_section_text
from researchlens.preprocessing.section_parser import split_into_sections

sample_text = """
Abstract
This study investigates transformer architectures for summarization.

Methodology
We sampled 500 academic articles (Table 1) and evaluated accuracy.
"""

sections = split_into_sections(sample_text, min_length=20)
for name, text in sections.items():
    cleaned = clean_section_text(text)
    print(f"[{name.upper()}] Cleaned: {cleaned}")
```
