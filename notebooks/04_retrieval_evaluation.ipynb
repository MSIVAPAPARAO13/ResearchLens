# 04. Retrieval Evaluation Benchmark

Evaluate dense and hybrid retrieval accuracy using Precision@K, Recall@K, and MRR metrics.

```python
from researchlens.evaluation.retrieval_metrics import precision_at_k, recall_at_k

retrieved = ["doc_1", "doc_3", "doc_5", "doc_2", "doc_4"]
relevant = ["doc_2", "doc_3"]

p_at_3 = precision_at_k(retrieved, relevant, k=3)
r_at_3 = recall_at_k(retrieved, relevant, k=3)
print(f"Precision@3: {p_at_3}, Recall@3: {r_at_3}")
```
